Q3. Set the random seed to your own roll number using random.seed(your_roll_number) before generating your numbers — this keeps your list random-looking but reproducible and unique to you (a different seed produces a different list, so this cannot be copied from a classmate). 

```python
import random
roll_number=1024170298
random.seed(roll_number)
```

i. Generate a list of 100 random numbers between 100 and 900 (inclusive). 

```python
numbers = [random.randint(100,900) for i in range(100)]
print(numbers)
```

ii. Count and print all odd numbers in the list. 

```python
odd_count=0
for num in numbers:
    if num%2!=0:
        print(num)
        odd_count+=1
print("Odd numbers: ",odd_count)
```

iii. Count and print all even numbers in the list.

```python
even_count=0
for num in numbers:
    if num%2==0:
        print(num)
        even_count+=1
print("Even numbers: ",even_count)
```

iv. Count and print all prime numbers in the list, and additionally build the actual list of prime numbers found using list comprehension. 

```python
def is_prime(n):
    if n<2:
        return False
    for i in range(2,int(n//2)+1):
        if n%i == 0:
            return False
    return True

prime_count=0
for num in numbers:
    if is_prime(num):
        print(num)
        prime_count+=1
print("Prime numbers: ",prime_count)
```

```python
# List Comprehension 
prime_list=[num for num in numbers if is_prime(num)]
print("Prime numbers: ",prime_list)
```

v. Print the number that occurs most frequently in your list, and how many times it occurs. 

```python
most_frequent = max(numbers,key = numbers.count)
print("Most Frequent: ",most_frequent)
print("Frequency: ",numbers.count(most_frequent))
```

Q4.  Using the same 8 digits of your roll number from Q1 (before multiplying by 10), create set A = {digit × 7 for each digit} and set B = {digit × 9 for each digit}. Print both sets.

i. Find the union of A and B — all unique values across both sets. 

```python
print("Union: ",A.union(B))
```

ii. Find the intersection of A and B — values common to both. 

```python
print("Intersection: ",A.intersection(B))
```

iii. Find A − B and B − A separately (not combined) using difference(), and explain in one line how this differs from symmetric_difference(). 

```python
print("A-B: ",A.difference(B))
print("B-A: ",B.difference(A))
```

iv. Find the symmetric difference of A and B. 

```python
print("Symmetric Difference:", A.symmetric_difference(B))
```

v. Check whether A is a subset of B, and whether B is a superset of A, using issubset() and issuperset(). 

```python
print("Is A subset of B?", A.issubset(B))
print("Is B superset of A?", B.issuperset(A))
```

vi. Ask the user to input a value X. Remove X from set A if present using discard() (not remove()) — explain in one line why discard() is safer than remove() when you're not sure the value exists.
