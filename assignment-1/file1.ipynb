Q1(d) Find the most frequent value and their indices in the following arrays

Q1(h) For the matrices

find inner, outer and cartesian product?

```python
x = np.array([[2,3,4],[3,2,9]])

y = np.array([[1,5,0],[5,10,3]])
```

```python
from itertools import product

x = np.array([[2,3,4],[3,2,9]])

y = np.array([[1,5,0],[5,10,3]])

print("Inner Product:")
print(np.inner(x, y))

print("\nOuter Product:")
print(np.outer(x, y))

print("\nCartesian Product:")
cart = np.array(list(product(x.flatten(), y.flatten())))
print(cart)
```

Q4(a) Write a function named img_to_array(path) that reads an image from a specified path and saves it as a text file on the local machine.

```python
from PIL import Image

def img_to_array(path):
    img = Image.open(path)

    arr = np.array(img)

    # RGB Image
    if len(arr.shape) == 3:
        print("RGB Image Detected")

        # Save each color channel separately
        np.savetxt("red_channel.txt", arr[:, :, 0], fmt="%d")
        np.savetxt("green_channel.txt", arr[:, :, 1], fmt="%d")
        np.savetxt("blue_channel.txt", arr[:, :, 2], fmt="%d")

    # Grayscale Image
    elif len(arr.shape) == 2:
        print("Grayscale Image Detected")
        np.savetxt("grayscale.txt", arr, fmt="%d")

    else:
        print("Unsupported Image Format")

    return arr
```
